```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Leitura do arquivo xlsx

df_analise = pd.read_excel("analise_funcionarios.xlsx")
display(df_analise)
```

```python
# Desmembrar a tabela

display(df_analise.shape)
display(df_analise.info())
display (df_analise.columns)
display(df_analise.describe())
```

```python
# Padronização

df_analise["Nome"] = df_analise["Nome"].str.title().str.strip()

df_analise["Departamento"] = df_analise["Departamento"].str.title().str.strip()

padronizar_departamentos = {
    "T.I": "TI",
    "T.i": "TI",
    "T.i.": "TI",
    "T.I.": "TI",
    "Ti": "TI",
    "Tecnologia da Informação": "TI",
    "Tecnologia Da Informação": "TI",
    "Rh": "RH",
    "Recursos Humanos": "RH"
}

df_analise["Departamento"] = df_analise["Departamento"].replace(padronizar_departamentos)

display(df_analise)
```

```python
# Tratamento de dados duplicados ou erros

duplicadas = df_analise.duplicated()
display(duplicadas)

total_duplicadas = df_analise.duplicated().sum()
display(total_duplicadas)

linhas_repetidas = df_analise[df_analise.duplicated()]
display(linhas_repetidas)

df_analise = df_analise.drop(df_analise[df_analise["Nome"] == "Xênia Henriques"].index)

df_analise = df_analise.drop_duplicates(subset=["Nome","Idade","Salario"])
display(df_analise)
```

```python
# Tratamento de linhas vazias em colunas string 

df_analise = df_analise.copy()

df_analise["Satisfacao"] = df_analise["Satisfacao"].fillna("Não informado")

df_analise["Motivo_Desligamento"] = df_analise["Motivo_Desligamento"].fillna("Funcionário Ativo")

display(df_analise)
```

```python
# Criação de novas colunas e posicionamento das colunas

def avaliar_desempenho(Desempenho):
    if pd.isna(Desempenho):
        return "Indisponível"

    elif Desempenho >= 9.0:
        return "Excelente Desempenho"
    
    elif Desempenho >= 7.0:
        return "Bom Desempenho"
    
    else:
        return "Desempenho Insatisfatório"


df_analise["Avaliação"] = df_analise["Desempenho"].apply(avaliar_desempenho)

def avaliar_salario(Salario):
    if Salario >= 10000.00:
        return "Salário Muito Alto"

    elif Salario >= 6000.00:
        return "Salário Alto"

    else:
        return "Salário Padrão"


df_analise["Faixa Salarial"] = df_analise ["Salario"].apply(avaliar_salario)

ordem_colunas = [
     "Nome", "Idade", "Departamento", "Salario", "Faixa Salarial", 
    "Desempenho", "Avaliação", "Satisfacao", "Motivo_Desligamento"
]

df_analise = df_analise [ordem_colunas]

display(df_analise)
```

```python
# Métricas

# .1 - MÉTRICAS SALÁRIO

# .1.1 - Salário Total

salario_total = df_analise["Salario"].sum()

# .1.2 - Salário Médio

salario_medio = df_analise["Salario"].mean() 

# .1.3 - Maior Salário

idx_maior_salario = df_analise["Salario"].idxmax()

nome_maior_salario = df_analise.loc[idx_maior_salario, "Nome"]

valor_maior_salario = df_analise.loc[idx_maior_salario, "Salario"]

# .1.4 - Menor Salário

idx_menor_salario = df_analise["Salario"].idxmin()

nome_menor_salario = df_analise.loc[idx_menor_salario, "Nome"]

valor_menor_salario = df_analise.loc[idx_menor_salario, "Salario"]
# ----------------------------------------------------------------------------------------------

# .2 - MÉTRICAS DESEMPENHO

# .2.1 - Desempenho Médio

media_desempenho = df_analise["Desempenho"].mean() 

# .2.2 - Maior Desempenho

idx_maior_desempenho = df_analise["Desempenho"].idxmax()

pessoa_maior_desempenho = df_analise.loc[idx_maior_desempenho, "Nome"]

nota_maior_desempenho = df_analise.loc[idx_maior_desempenho, "Desempenho"]

# .2.3 - Menor Desempenho

idx_menor_desempenho = df_analise["Desempenho"].idxmin()

pessoa_menor_desempenho = df_analise.loc[idx_menor_desempenho, "Nome"]

nota_menor_desempenho = df_analise.loc[idx_menor_desempenho, "Desempenho"]
# ----------------------------------------------------------------------------------------------

# .3 - MÉTRICAS IDADE

# .3.1 - Média de Idade

média_idade = df_analise["Idade"].mean()

# .3.2 - Maior Idade

idx_maior_idade = df_analise["Idade"].idxmax()

nome_maior_idade = df_analise.loc[idx_maior_idade, "Nome"]

numero_maior_idade = df_analise.loc[idx_maior_idade, "Idade"]

# .3.3 - Menor Idade

idx_menor_idade = df_analise["Idade"].idxmin()

nome_menor_idade = df_analise.loc[idx_menor_idade, "Nome"]

numero_menor_idade = df_analise.loc[idx_menor_idade, "Idade"]
# ----------------------------------------------------------------------------------------------

# .4 - MÉTRICAS DEPARTAMENTO

# .4.1 - Maior Departamento

contagem_departamento = df_analise["Departamento"].value_counts()

nome_maior_departamento = contagem_departamento.idxmax()

qtd_maior_departamento = contagem_departamento.max()

# .4.2 - Menor Departamento

contagem_departamento = df_analise["Departamento"].value_counts()

nome_menor_departamento = contagem_departamento.idxmin()

qtd_menor_departamento = contagem_departamento.min()
# ----------------------------------------------------------------------------------------------

# Resumo

df_resumo_geral = pd.DataFrame({
    "Indicador": [
        "Salário Total", 
        "Salário Médio",
        "Maior Salário",
        "Menor Salário",
        "Desempenho Médio", 
        "Maior Desempenho", 
        "Menor Desempenho", 
        "Média de Idade", 
        "Pessoa Mais Velha", 
        "Pessoa Mais Nova",
        "Maior Departamento",
        "Menor Departamento"
    ],
    "Resultado": [
        f"R$ {salario_total:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."),
        f"R$ {salario_medio:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."),
        f"R$ {valor_maior_salario:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."),
        f"R$ {valor_menor_salario:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."),
        f"{media_desempenho:.2f}".replace(".", ","),
        f"{pessoa_maior_desempenho} ({f'{nota_maior_desempenho:.2f}'.replace('.', ',')})",
        f"{pessoa_menor_desempenho} ({f'{nota_menor_desempenho:.2f}'.replace('.', ',')})",
        f"{média_idade:.1f} anos".replace(".", ","),
        f"{nome_maior_idade} ({int(numero_maior_idade)} anos)",
        f"{nome_menor_idade} ({int(numero_menor_idade)} anos)",
        f"{nome_maior_departamento} ({int(qtd_maior_departamento)} funcionários)",
        f"{nome_menor_departamento} ({int(qtd_menor_departamento)} funcionários)"
    ]
})
display(df_resumo_geral)
```

```python
# Planilha Final

display(
    df_analise.style.format(
        {
        "Desempenho": "{:.1f}",
        "Idade": "{:.0f}",
        "Salario": lambda x: f"R$ {x:,.2f}".replace(",", "X").replace(".", ",").replace("X", ".") if pd.notna(x) else "Não informado"
        },  
    )
)

display(df_resumo_geral.style)
```

```python
# Exportação pro Excel

df_analise_estilizado = df_analise.style.format(
    {
        "Desempenho": "{:.1f}",
        "Idade": "{:.0f}",
        "Salario": lambda x: f"R$ {x:,.2f}".replace(",", "X").replace(".", ",").replace("X", ".") if pd.notna(x) else x
    }
)


nome_arquivo = "Analise_Final_Tratada.xlsx"

with pd.ExcelWriter(nome_arquivo) as writer:
    df_analise_estilizado.to_excel(writer, sheet_name="Dados Tratados", index=False, na_rep="Não Informado")
    df_resumo_geral.to_excel(writer, sheet_name="Resumo Indicadores", index=False)

print(f"\n Sucesso! O arquivo '{nome_arquivo}' foi gerado e as células vazias agora dizem 'Não Informado' no Excel.")
```
